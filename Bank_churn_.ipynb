***Outlier Treatment***: IQR treatment

Age:

```python
Age_non_churned=bank_raw_data[bank_raw_data['Exited']==0]['Age']
```

```python
Age_non_churned_q25=np.percentile(Age_non_churned,25)
Age_non_churned_q75=np.percentile(Age_non_churned,75)
Age_non_churned_IQR=Age_non_churned_q75-Age_non_churned_q25
Age_non_churned_upper=Age_non_churned_q75+1.5*Age_non_churned_IQR
Age_non_churned_lower=Age_non_churned_q25-1.5*Age_non_churned_IQR
```

```python
bank_raw_data=bank_raw_data.drop(bank_raw_data[bank_raw_data['Age']>Age_non_churned_upper].index)
```

```python
# detection of outliers in estimated salary using box plot:
sns.boxplot(data=bank_raw_data,x='Exited',y='Age')
plt.title('Age Vs Exited')
```

***Feature Scaling:*** 

We need to scale the feature which includes continuous values with large interval into smaller interval using ROBUST SCALING. The features we need to scale includes "Credit score", "Balance", and "Estimated Salary".

```python
from sklearn.preprocessing import RobustScaler
rob_sca=RobustScaler()

#Scaling variables:
bank_raw_data['CreditScore_scaled']=rob_sca.fit_transform(bank_raw_data['CreditScore'].values.reshape(-1,1))
bank_raw_data['Balance_scaled']=rob_sca.fit_transform(bank_raw_data['Balance'].values.reshape(-1,1))
bank_raw_data['EstimatedSalary_scaled']=rob_sca.fit_transform(bank_raw_data['EstimatedSalary'].values.reshape(-1,1))
```

```python
bank_raw_data.columns
```

```python
bank_raw_data.describe()
```

```python
#Removing non-scaled fetures:
bank_raw_data.drop(["CreditScore","Balance","EstimatedSalary"],axis=1,inplace=True)
```

```python
bank_raw_data.columns
```

```python
#positioning the scaled feature:
Scaled_CreditScore=bank_raw_data['CreditScore_scaled']
Scaled_Balance=bank_raw_data['Balance_scaled']
Scaled_EstimatedSalary=bank_raw_data['EstimatedSalary_scaled']

bank_raw_data.insert(3,"CreditScore",Scaled_CreditScore)
bank_raw_data.insert(8,"Balance",Scaled_Balance)
bank_raw_data.insert(12,"EstimatedSalary",Scaled_Balance)
```

```python
#Removing Duplicate Columns:
bank_raw_data.drop(['CreditScore_scaled','Balance_scaled', 'EstimatedSalary_scaled'],axis=1,inplace=True)
```

```python
# Now handling the imbalanced data with SMOTE

from imblearn.over_sampling import RandomOverSampler

ros = RandomOverSampler(random_state=42)
X_resampled, y_resampled = ros.fit_resample(X, y)


```

```python
# Splitting the dataset into Training and Testing Data
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X_resampled,y_resampled,test_size=0.2, random_state = 42)
```

```python
# Standardizing the Dataset

from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```
