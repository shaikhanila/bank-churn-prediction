## Logistic Regression:

***Data Splitting:*** 80-20 split

***Without Library:***

```python
import numpy as np

def sigmoid(z):
    z = np.array(z, dtype=np.float64)  # Ensure z is a NumPy array
    return 1 / (1 + np.exp(-z))

def train_logistic_regression(X, y, lr=0.01, epochs=500):
    X = np.array(X, dtype=np.float64)  # Ensure X is a NumPy array
    y = np.array(y, dtype=np.float64)  # Ensure y is a NumPy array
    m, n = X.shape
    weights = np.zeros(n)
    bias = 0

    for _ in range(epochs):
        linear_model = np.dot(X, weights) + bias
        y_pred = sigmoid(linear_model)

        # Compute gradients
        dw = (1/m) * np.dot(X.T, (y_pred - y))
        db = (1/m) * np.sum(y_pred - y)

        # Update parameters
        weights -= lr * dw
        bias -= lr * db

    return weights, bias

def predict_logistic_regression(X, weights, bias):
    X = np.array(X, dtype=np.float64)  # Ensure X is a NumPy array
    linear_model = np.dot(X, weights) + bias
    y_pred = sigmoid(linear_model)
    return (y_pred > 0.5).astype(int)
```

```python
# Train model
weights, bias = train_logistic_regression(X_train.to_numpy(), Y_train.to_numpy().ravel())
```

```python
Y_pred = predict_logistic_regression(X_test.to_numpy(), weights, bias)
```

***Comparing both Logistic Regression:***

```python
def evaluate_model(y_true, y_pred, model_name):
    accuracy = accuracy_score(y_true, y_pred)
    precision = precision_score(y_true, y_pred)
    recall = recall_score(y_true, y_pred)
    f1 = f1_score(y_true, y_pred)
    
    print(f"\n📊 {model_name} Performance:")
    print(f"🔹 Accuracy: {accuracy*100:.2f}")
    print(f"🔹 Precision: {precision*100:.2f}")
    print(f"🔹 Recall: {recall*100:.2f}")
    print(f"🔹 F1 Score: {f1*100:.2f}")
    
    return accuracy, precision, recall, f1
```

##  Feature Engineering and Selection: Using PCA Transformation

```python
from sklearn.decomposition import PCA
pca=PCA()
```

```python
#drop the target variable:
bank_data_without_target=bank_data.drop('Exited',axis=1)
target_variable=bank_data['Exited']
```

```python
#applying PCA transformation:
pca.fit(bank_data_without_target)

#calculating thr variance ratio:
variance_ratio=pca.explained_variance_ratio_
variance_ratio_cumulative_sum=np.cumsum(variance_ratio)

#Ploting Elbow graph for selection of number of feature:
plt.figure(figsize=(8, 5))
plt.plot(range(1, len(variance_ratio) + 1), variance_ratio_cumulative_sum, marker='o', linestyle='--', label='Cumulative Variance')
plt.xlabel('Number of Principal Components')
plt.ylabel('Variance cumulative sum')
plt.title('Elbow Method')
plt.grid()

# Ploting elbow on the graph:
elbow = np.argmax(np.diff(variance_ratio_cumulative_sum) < 0.01) + 1
plt.axvline(elbow, color='r', linestyle='--', label=f'Elbow at {elbow}')
plt.legend()
```

From the PCA transformation we can convert all the feature into 3 principal feature.

***Applying PCA transformation:***

```python
pca=PCA(n_components=elbow)
bank_data_PCA_raw=pca.fit_transform(bank_data_without_target)

#Converting pca transformered data into data set:
bank_data_PCA=pd.DataFrame(bank_data_PCA_raw,columns=[f'PC{i+1}' for i in range (bank_data_PCA_raw.shape[1])])
bank_data_PCA['Exited']=target_variable.values
```
